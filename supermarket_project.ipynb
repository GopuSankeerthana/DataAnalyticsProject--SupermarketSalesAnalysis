```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import plotly
import plotly.express as px

import plotly.graph_objects as go
```

```python
df["Date"] = pd.to_datetime(df["Date"])    # converts date column from string to real time date format
```

```python
df["year"] = df["Date"].dt.year
df["month"] = df["Date"].dt.month #creating time features
df["Date"] = df["Date"].dt.day
```

```python
#recheck if there are any null values after feature extraction
df.isnull().sum()
df.duplicated().sum()
```

```python
df["Hour"] = pd.to_datetime(df["Time"]).dt.hour

def slot(h):
    if h < 12: return "Morning"
    elif h < 17: return "Afternoon"
    else: return "Evening"

df["Time_slot"] = df["Hour"].apply(slot)
```

```python
sales_branch = df.groupby("Branch")["Total"].sum()
sales_branch
```

```python
df["Product line"] = df["Product line"].str.split(" and ")
df = df.explode("Product line")
```

```python
sales_product = df.groupby("Product line")["Total"].sum().sort_values()
sales_product
```

```python
sales_month = df.groupby(["month" , "Product line"])["Total"].sum()
sales_month
```

```python
sales_gender = df.groupby(["Gender", "Product line"])["Total"].sum()
sales_gender
```

```python
payment_counts = df["Payment"].value_counts()
payment_counts
```

```python
rating_product = df.groupby("Product line")["Rating"].mean().sort_values()
rating_product
```

```python
sales_month_category = df.groupby(
    ["month", "Product line"]
)["Total"].sum().reset_index()

fig1 = px.bar(
    sales_month_category,
    x="month",
    y="Total",
    color="Product line",
    barmode="group",
    title="Sales per Month by Category",
    text_auto=True,
    template="plotly_dark"
)

fig1.show()

```

```python
cat_rev = df.groupby("Product line")["Total"].sum().reset_index()

fig2 = px.bar(
    cat_rev,
    x="Product line",
    y="Total",
    color="Total",
    title="Revenue by Product Category",
    text_auto=True,
    template="plotly_dark"
)

fig2.show()
```

```python
branch_rev = df.groupby("Branch")["Total"].sum().reset_index()

fig3 = px.bar(
    branch_rev,
    x="Branch",
    y="Total",
    color="Branch",
    text_auto=True,
    title="Revenue by Branch",
    template="plotly_dark"
)

fig3.show()
```

```python
fig4 = px.pie(
    df,
    names="Payment",
    title="Payment Method Share",
    template="plotly_dark"
)

fig4.show()
```

```python
month_rev = df.groupby("month")["Total"].sum().reset_index()

fig5 = px.line(
    month_rev,
    x="month",
    y="Total",
    markers=True,
    title="Monthly Revenue Trend",
    template="plotly_dark"
)

fig5.show()
```

```python
fig6 = px.histogram(
    df,
    x="Rating",
    nbins=20,
    title="Customer Rating Distribution",
    template="plotly_dark",
    text_auto=True
)

fig6.show()
```

```python
slot_rev = df.groupby("Time_slot")["Total"].sum().reset_index()

fig7 = px.bar(
    slot_rev,
    x="Time_slot",
    y="Total",
    color="Time_slot",
    text_auto=True,
    title="Revenue by Time Slot",
    template="plotly_dark"
)
fig7.show()
```
